# Customer Churn Analysis

## Business Problem
Analyze customer churn behavior and identify factors affecting customer retention using exploratory data analysis and visualization techniques.

# 🎯 Project Objective

The goal of this project is to analyze customer churn behavior and identify factors contributing to customer attrition.

Using exploratory data analysis (EDA), this project uncovers trends related to customer contracts, tenure, and monthly charges to help businesses improve customer retention strategies.

## 🛠 Technologies Used

- Python
- Pandas
- Matplotlib
- Seaborn
- Jupyter Notebook

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df.info()
df.head()
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Churn'].value_counts()
```

```python
sns.countplot(x='Churn', data=df)
plt.title("Distribution of Customer Churn")
plt.show()
```

# 📌 Key Insights

- Majority of customers remained with the company

- A significant number of customers still churned, indicating retention challenges

- Customer churn represents a potential business risk impacting long-term revenue

- Retention strategies could help reduce customer loss

```python
sns.countplot(x='Contract', hue='Churn', data=df)
plt.title("Customer Churn Across Contract Types")
plt.xticks(rotation=45)
plt.show()
```

# 📌 Key Insights

- Month-to-month customers showed highest churn rates

- Customers with long-term contracts had better retention

- Long-term subscriptions improve customer loyalty

```python
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.title("Relationship Between Monthly Charges and Customer Churn")
plt.show()
```

```python
sns.histplot(data=df, x='tenure', hue='Churn', multiple='stack')
plt.title("Tenure Distribution")
plt.show()
```

# 📌 Key Insights

- Customers with shorter tenure were more likely to churn

- Long-term customers showed better retention rates

- Churn probability decreases as customer tenure increases

- Customer loyalty appears stronger after long-term engagement

# 📌 Business Recommendations

- Encourage customers to move to long-term contracts

- Offer retention incentives for high-risk customers

- Provide loyalty benefits for new customers during early subscription periods

- Introduce pricing strategies to reduce churn among customers with high monthly charges

# 💼 Business Value

This analysis can help telecom companies:

- Reduce customer churn
- Improve retention strategies
- Identify high-risk customers
- Increase customer lifetime value
- Optimize pricing and contract plans

# 📌 Final Conclusion

This project analyzed customer churn behavior using the Telco Customer Churn dataset.

Key findings include:

- Customers with month-to-month contracts showed significantly higher churn rates
- Customers with shorter tenure were more likely to leave
- Higher monthly charges increased churn probability
